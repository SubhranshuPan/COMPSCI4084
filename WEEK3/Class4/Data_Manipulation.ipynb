```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('shopping.csv', parse_dates=['Date'], dayfirst=True)
df.head()
```

```python
df['Price'] = df['Price'].fillna(0)
df['Quantity'] = df['Quantity'].fillna(0)
```

```python
df['Total'] = df['Price'] * df['Quantity']
```

```python
df
```

```python
product_sales = df.groupby('Product')['Total'].sum()
product_sales.plot(kind='bar', title='Total sales per product', color=['red', 'yellow', 'orange'])
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Date'].dt.to_period("M")
monthly_sales = df.groupby("Month")["Total"].sum()

monthly_sales.plot(kind="line", title="Total sales over time", marker="o")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```
